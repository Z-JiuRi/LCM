```python
# 读取所有图片，找到最大和最小分辨率
import os
max_res = (0, 0)
min_res = (np.inf, np.inf)
data_dir = '/home/zxd/zxd/Huawei/datasets/WAIRD/data/scenario_2'
for folder_name in os.listdir(data_dir):
    folder_path = os.path.join(data_dir, folder_name)
    if os.path.isdir(folder_path):
        for file_name in os.listdir(folder_path):
            if file_name.endswith('.png'):
                file_path = os.path.join(folder_path, file_name)
                img = cv2.imread(file_path)
                h, w, _ = img.shape
                max_res = (max(max_res[0], h), max(max_res[1], w))
                min_res = (min(min_res[0], h), min(min_res[1], w))
                print(f'{file_path}: {img.shape}')
print(f'max_res={max_res}, min_res={min_res}')
```
